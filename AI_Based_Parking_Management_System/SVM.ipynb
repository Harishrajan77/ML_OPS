```python
from sklearn.svm import SVC
from sklearn.model_selection import train_test_split
import pandas as pd
```

```python
df = pd.read_csv('parking_dataset.csv')
df['Parking_Status'] = (
    df['Occupancy_Rate'] > 60 ).astype(int)
df.head()
```

```python
x = df[['Hour','Vehicle_Count','Entry_Count','Exit_Count','Peak_Hour', 'Temperature']]
y = df['Parking_Status']
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
model = SVC()
model.fit(x_train,y_train)
print(model.score(x_test,y_test))
```

```python
model_c = SVC(C = 1.0)
model_c.fit(x_train,y_train)
print(model_c.score(x_test,y_test))
```

```python
model_r = SVC(kernel='rbf')
model_r.fit(x_train,y_train)
print(model_r.score(x_test,y_test))
```

```python
model_l = SVC(kernel='linear')
model_l.fit(x_train,y_train)
print(model_l.score(x_test,y_test))
```

```python
model_g = SVC(gamma='scale')
model_g.fit(x_train,y_train)
print(model_g.score(x_test,y_test))
```
