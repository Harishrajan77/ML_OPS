```python
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB,MultinomialNB
import pandas as pd
```

```python
df = pd.read_csv('parking_dataset.csv')
```

```python
df = pd.read_csv('parking_dataset.csv')
df['Parking_Status'] = (
    df['Occupancy_Rate'] > 60   
).astype(int)
df.head()
```

```python
x = df[[
 'Hour',
 'Vehicle_Count',
 'Entry_Count',
 'Exit_Count',
 'Peak_Hour',
 'Temperature'
]]
y = df['Parking_Status']
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
print(x_train.shape, x_test.shape, y_train.shape, y_test.shape)
```

```python
model_1 = GaussianNB()
model_1.fit(x_train, y_train)
predicted_1 = model_1.predict(x_test)
print('Actual Values:', y_test.values)
print('Predicted Values:', predicted_1)
print('GaussianNB Accuracy:', model_1.score(x_test, y_test))
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)

model_2 = MultinomialNB()
model_2.fit(x_train_scaled, y_train)

predicted_2 = model_2.predict(x_test_scaled)

print("Actual Values:", y_test.values)
print("Predicted Values:", predicted_2)
print("MultinomialNB Accuracy:",model_2.score(x_test_scaled, y_test))
```
